# MiniMind 数据集抽样查看

每个数据集一个 section，修改该 section 里的 `N` 后运行即可。默认 `N = 5`。

先运行下面的 setup cell（`SEED` / `RANDOM_SAMPLE` 在里面调）。输出是每条样本的原始 JSON 结构。

```python
import json, os, random
from pathlib import Path

DATASET_DIR = Path("..") / "dataset"
SEED = None          # 设成整数可复现；None = 每次运行都不同
RANDOM_SAMPLE = True # True: 随机抽样；False: 取文件开头的前 N 条

def load_samples(name, n=5):
    """返回原始 JSON 对象的 list。不把整个文件读进内存：随机跳到某个字节偏移，读取其后的下一整行。"""
    path = DATASET_DIR / f"{name}.jsonl"
    if not RANDOM_SAMPLE:
        with open(path, encoding="utf-8") as f:
            return [json.loads(line) for _, line in zip(range(n), f)]
    size = os.path.getsize(path)
    rng = random.Random(SEED)
    offsets = sorted(rng.randrange(size) for _ in range(n))
    out = []
    with open(path, "rb") as f:
        for off in offsets:
            f.seek(off)
            f.readline()                 # 丢弃可能被截断的半行
            line = f.readline()
            if not line:                 # 落在末尾，回到文件开头
                f.seek(0); line = f.readline()
            out.append(json.loads(line))
    return out
```

## pretrain_t2t

预训练全量 (`text`)

```python
N = 5
samples = load_samples("pretrain_t2t", N)
for i, ex in enumerate(samples, 1):
    print(f"--- #{i} ---")
    print(json.dumps(ex, ensure_ascii=False, indent=2))
```
